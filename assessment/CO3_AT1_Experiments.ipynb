TASK 1: MD5 and SHA-256 Hashing Implementation & Analysis

```python
import hashlib
import time

def compare_hashing_algorithms():
    messages = [
        b"Hello, Cryptography and Network Security!",
        b"Data integrity and authentication are critical components of secure systems.",
        b"A" * 5000,  # 5 KB block size test
        b"Simats Engineering - CSA51 Assessment"
    ]

    print("=== TASK 1: MD5 vs SHA-256 Hashing Comparison ===\n")

    for i, msg in enumerate(messages, 1):
        print(f"Message {i} (Length: {len(msg)} bytes): {msg[:30]}..." if len(msg) > 30 else f"Message {i}: {msg}")

        # MD5 Hashing
        start = time.perf_counter()
        md5_digest = hashlib.md5(msg).hexdigest()
        md5_time = (time.perf_counter() - start) * 1000  # ms

        # SHA-256 Hashing
        start = time.perf_counter()
        sha256_digest = hashlib.sha256(msg).hexdigest()
        sha256_time = (time.perf_counter() - start) * 1000  # ms

        print(f"  [MD5]    Digest: {md5_digest} ({md5_time:.4f} ms)")
        print(f"  [SHA-256] Digest: {sha256_digest} ({sha256_time:.4f} ms)\n")

if __name__ == "__main__":
    compare_hashing_algorithms()
```

TASK 2: RSA Digital Signature Scheme

```python
import rsa

def rsa_signature_demo():
    print("=== TASK 2: RSA Digital Signature Demonstration ===\n")

    # 1. Key Generation
    public_key, private_key = rsa.newkeys(512)
    print("[*] RSA Key Pair Generated.")

    # 2. Message Signing
    message = b"Authorize financial transaction of $10,000 to Account #4582."
    signature = rsa.sign(message, private_key, 'SHA-256')
    print(f"Original Message: {message.decode()}")
    print(f"Generated Signature (Hex): {signature.hex()}\n")

    # 3. Successful Verification
    try:
        rsa.verify(message, signature, public_key)
        print("[+] Verification SUCCESS: Message is authentic and untampered.")
    except rsa.VerificationError:
        print("[-] Verification FAILED.")

    # 4. Tampered Message Test
    tampered_message = b"Authorize financial transaction of $100,000 to Account #4582."
    print(f"\nTried Tampered Message: {tampered_message.decode()}")
    try:
        rsa.verify(tampered_message, signature, public_key)
        print("[+] Verification SUCCESS.")
    except rsa.VerificationError:
        print("[-] Verification FAILED as expected! Data tampering detected.")

if __name__ == "__main__":
    rsa_signature_demo()
```

TASK 3: HMAC Simulation vs. Simple Hash Integrity Check

```python
import hashlib
import hmac

def hmac_vs_simple_hash():
    print("=== TASK 3: HMAC vs Simple Hash Integrity Check ===\n")

    secret_key = b"EnterpriseSecretKey99"
    message = b"Command: Execute system backup sequence."

    # 1. Simple Hash-based check (Naive concatenation: Hash(Key + Message))
    simple_hash = hashlib.sha256(secret_key + message).hexdigest()
    print(f"Simple Hash (Key || Msg): {simple_hash}")

    # 2. Secure HMAC using RFC 2104 standards
    hmac_obj = hmac.new(secret_key, message, hashlib.sha256)
    hmac_signature = hmac_obj.hexdigest()
    print(f"HMAC Signature (SHA-256): {hmac_signature}\n")

    print("[Analysis Summary]:")
    print(" - Naive Hash concatenation (Key + Message) using Merkle-Damgard hashes is vulnerable to Length Extension Attacks.")
    print(" - HMAC uses nested inner/outer padding layers, protecting the secret key and rendering length extension attacks impossible.")

if __name__ == "__main__":
    hmac_vs_simple_hash()
```

TASK 4: Simplified Kerberos Authentication & Replay Prevention

```python
import time
import hashlib
import hmac

class KerberosKDC:
    def __init__(self):
        self.secret_key = "ServiceSecretMasterKey"

    def issue_ticket(self, client_id):
        # Ticket contains client ID and expiration timestamp (valid for 30s)
        expiry = time.time() + 30
        ticket_data = f"{client_id}:{expiry}".encode()
        sig = hmac.new(self.secret_key.encode(), ticket_data, hashlib.sha256).hexdigest()
        return {"ticket": ticket_data, "signature": sig}

class ApplicationServer:
    def __init__(self, secret_key):
        self.secret_key = secret_key

    def validate_ticket(self, client_id, ticket_dict):
        ticket_data = ticket_dict["ticket"]
        expected_sig = hmac.new(self.secret_key.encode(), ticket_data, hashlib.sha256).hexdigest()

        # Verify HMAC signature
        if not hmac.compare_digest(expected_sig, ticket_dict["signature"]):
            return False, "Invalid Ticket Signature"

        # Check Expiration (Replay Prevention)
        _, expiry_str = ticket_data.decode().split(":")
        if time.time() > float(expiry_str):
            return False, "Ticket Expired / Potential Replay Attack Detected"

        return True, "Authentication Successful"

def run_kerberos_simulation():
    print("=== TASK 4: Simplified Kerberos Simulation ===\n")
    kdc = KerberosKDC()
    server = ApplicationServer("ServiceSecretMasterKey")

    client = "user_alpha"
    ticket = kdc.issue_ticket(client)

    print("[*] Client requested and received ticket from KDC.")

    # Valid immediate request
    success, msg = server.validate_ticket(client, ticket)
    print(f"Immediate Validation: {msg}")

    # Replay Simulation (Simulate expired time window)
    print("\n[*] Simulating Replay Attack after expiration window...")
    ticket["ticket"] = f"{client}:{time.time() - 10}".encode() # force expired timestamp
    success, msg = server.validate_ticket(client, ticket)
    print(f"Replay Attempt Validation: {msg}")

if __name__ == "__main__":
    run_kerberos_simulation()
```

TASK 5: ElGamal Digital Signature Scheme

```python
import random
import hashlib

def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

def mod_inverse(a, m):
    # Extended Euclidean Algorithm to find modular inverse of a mod m
    m0 = m
    y = 0
    x = 1
    if m == 1:
        return 0
    while a > 1:
        # q is quotient
        q = a // m
        t = m
        # m is remainder now, process same as Euclid's algo
        m = a % m
        a = t
        t = y
        # Update y and x
        y = x - q * y
        x = t
    # Make x positive
    if x < 0:
        x = x + m0
    return x

class ElGamalSignature:
    def __init__(self):
        # Public parameters (Small prime for demo purposes)
        self.p = 467
        self.g = 2
        self.x = 127  # Private key
        self.y = pow(self.g, self.x, self.p)  # Public key component

    def sign(self, message):
        h = int(hashlib.sha256(message).hexdigest(), 16) % (self.p - 1)
        while True:
            k = random.randint(1, self.p - 2)
            if gcd(k, self.p - 1) == 1:
                break
        r = pow(self.g, k, self.p)
        k_inv = mod_inverse(k, self.p - 1)
        s = ((h - self.x * r) * k_inv) % (self.p - 1)
        return (r, s)

    def verify(self, message, signature):
        r, s = signature
        if not (0 < r < self.p):
            return False
        h = int(hashlib.sha256(message).hexdigest(), 16) % (self.p - 1)
        left = (pow(self.y, r, self.p) * pow(r, s, self.p)) % self.p
        right = pow(self.g, h, self.p)
        return left == right

def run_elgamal_demo():
    print("=== TASK 5: ElGamal Digital Signature Simulation ===\n")
    signer = ElGamalSignature()
    message = b"Classified cryptographic transmission."

    signature = signer.sign(message)
    print(f"Message: {message.decode()}")
    print(f"ElGamal Signature Pair (r, s): {signature}")

    is_valid = signer.verify(message, signature)
    print(f"Signature Verification Status: {is_valid}")

if __name__ == "__main__":
    run_elgamal_demo()
```
